Matching differently formatted County names on FIPS

```python
# Step 1: Create lookup from election_pivot: FIPS -> County
fips_to_county = election_pivot[['FIPS Code', 'County']].drop_duplicates().set_index('FIPS Code')['County']

# Step 2: Identify conflicting rows
conflict_mask = (
    unemployment['FIPS Code'].isin(fips_to_county.index) &
    (unemployment['County'] != unemployment['FIPS Code'].map(fips_to_county))
)

# Step 3: Apply correction — use election_pivot county name where FIPS matches but name differs
unemployment.loc[conflict_mask, 'County'] = unemployment.loc[conflict_mask, 'FIPS Code'].map(fips_to_county)
```

```python
education = pd.read_csv("/Users/matthewcavanaugh/Desktop/GitHub/us-presidential-elections-by-county/sources/Education CLEANED.csv")
```

```python
education = education.sort_values(by=['Year', 'State', 'County', 'FIPS Code'])
education.reset_index(inplace=True)
education.drop('index', axis=1, inplace=True)
education
```

```python
education['FIPS Code'] = education['FIPS Code'].astype(str).str.zfill(5)
education
```

```python
# Step 1: Create lookup from election_pivot: FIPS -> County
fips_to_county = election_pivot[['FIPS Code', 'County']].drop_duplicates().set_index('FIPS Code')['County']

# Step 2: Identify conflicting rows
conflict_mask = (
    education['FIPS Code'].isin(fips_to_county.index) &
    (education['County'] != education['FIPS Code'].map(fips_to_county))
)

# Step 3: Apply correction — use election_pivot county name where FIPS matches but name differs
education.loc[conflict_mask, 'County'] = education.loc[conflict_mask, 'FIPS Code'].map(fips_to_county)
```

```python
education
```

```python
education.groupby('Year')[['State', 'County', 'FIPS Code']].nunique()
```

```python
# Define the key columns to compare
key_cols = ['Year', 'State', 'County', 'FIPS Code']

# Find rows in education that are not in election_pivot
education_not_in_election = education.merge(
    election_pivot[key_cols].drop_duplicates(),
    on=key_cols,
    how='left',
    indicator=True
).query('_merge == "left_only"').drop(columns=['_merge'])

# Show the unique counties that are missing
missing_counties = education_not_in_election[['Year', 'State', 'County', 'FIPS Code']].drop_duplicates()

print(missing_counties.sort_values(by=['Year', 'State', 'County']))
```

We've already explored problems with Bedford City, VA and Broomfield County, CO. Spelling mistakes and duplicated codes in DC are here again.

Clifton Forge, VA, is another former independent city that gave up its status, doing so in 2001. The dataset seems to contain a "ghost row" for it in 2004. Let's drop it.

But here we see another complication with Kalawao County, Hawaii. It was founded to resettle leprosy patients, and to this day non-residents may only visit with permission. The only land access is by mule. Though technically named a county, it does not have the powers of one, and is governed by Maui County. Thus it does not even appear in our initial election dataset. With a population of 82, it may be safe to drop.

```python
education = education[~education['County'].str.contains("Broomfield County", na=False)]
education = education[~education['County'].str.contains("Bedford city", na=False)]
education = education[~education['County'].str.contains("Kalawao", na=False)]
education = education[~education['County'].str.contains("Clifton Forge city", na=False)]

education = education.replace({"District of Columbia": "District Of Columbia"})
```

```python
# Define the key columns to compare
key_cols = ['Year', 'State', 'County', 'FIPS Code']

# Find rows in education that are not in election_pivot
education_not_in_election = education.merge(
    election_pivot[key_cols].drop_duplicates(),
    on=key_cols,
    how='left',
    indicator=True
).query('_merge == "left_only"').drop(columns=['_merge'])

# Show the unique counties that are missing
missing_counties = education_not_in_election[['Year', 'State', 'County', 'FIPS Code']].drop_duplicates()

print(missing_counties.sort_values(by=['Year', 'State', 'County']))
```

```python
population = pd.read_csv("/Users/matthewcavanaugh/Desktop/GitHub/us-presidential-elections-by-county/sources/Population CLEANED.csv")
population
```

```python
pop_keys = population[['State', 'County']].copy()
elec_keys = election_pivot[['State', 'County']].copy()

# Normalize by stripping whitespace and converting to lowercase for comparison only
pop_keys['State_norm'] = pop_keys['State'].str.strip().str.lower()
pop_keys['County_norm'] = pop_keys['County'].str.strip().str.lower()

elec_keys['State_norm'] = elec_keys['State'].str.strip().str.lower()
elec_keys['County_norm'] = elec_keys['County'].str.strip().str.lower()

pop_set = set(pop_keys[['State_norm', 'County_norm']].apply(tuple, axis=1))
elec_set = set(elec_keys[['State_norm', 'County_norm']].apply(tuple, axis=1))

only_in_population = pop_set - elec_set
only_in_election = elec_set - pop_set

print("In population but NOT in election_pivot:")
for state, county in sorted(only_in_population):
    # Get original names from population dataframe (case sensitive)
    original = pop_keys[(pop_keys['State_norm'] == state) & (pop_keys['County_norm'] == county)]
    print(original[['State', 'County']].values[0])

print("\nIn election_pivot but NOT in population:")
for state, county in sorted(only_in_election):
    # Get original names from election_pivot dataframe
    original = elec_keys[(elec_keys['State_norm'] == state) & (elec_keys['County_norm'] == county)]
    print(original[['State', 'County']].values[0])
```
