# Interpolate values for all other years
We will estimate 2001-2009 values based on 2000 and 2010 actual values, then estimate 2011-2020 based on 2010 and 2021 actual values. Then we will reduce to only the years we need.

```python
import pandas as pd

numeric_cols = [
    'At least some college',
    "Bachelor's degree or higher",
    'High school graduate',
    'Less than high school'
]

def interpolate_piecewise(df):
    # For each county (FIPS), interpolate in two segments
    
    def interp_group(group):
        group = group.sort_values('Year').set_index('Year')
        
        # Part 1: 2000-2010
        part1 = group.loc[2000:2010].copy()
        part1 = part1.reindex(range(2000, 2011))
        # Fill FIPS Code and other static columns BEFORE interpolation
        part1['FIPS Code'] = group['FIPS Code'].iloc[0]
        part1['State'] = group['State'].iloc[0]
        part1['County'] = group['County'].iloc[0]
        part1[numeric_cols] = part1[numeric_cols].interpolate(method='linear')
        
        # Part 2: 2010-2021
        part2 = group.loc[2010:2021].copy()
        part2 = part2.reindex(range(2010, 2022))
        part2['FIPS Code'] = group['FIPS Code'].iloc[0]
        part2['State'] = group['State'].iloc[0]
        part2['County'] = group['County'].iloc[0]
        part2[numeric_cols] = part2[numeric_cols].interpolate(method='linear')
        
        # Combine, drop duplicated 2010 in part2
        part2 = part2.loc[2011:]
        combined = pd.concat([part1, part2])
        
        combined = combined.reset_index()
        return combined
    
    result = df.groupby('FIPS Code', group_keys=False).apply(interp_group)
    return result

# Apply
education_interpolated_piecewise = interpolate_piecewise(education_pivoted)
education_pivoted = education_interpolated_piecewise
```
