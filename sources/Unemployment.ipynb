```python
# Clean up text for aesthetic purposes
def clean_text(text):
    if isinstance(text, str):
        return text.title()  # Capitalizes the first letter of each word and makes the rest lowercase
    return text  # Return text unchanged if it's not a string

unemployment['County'] = unemployment['County'].apply(clean_text)
```

```python
unemployment[unemployment['County'].str.contains('St. Louis')]
```

```python
import pandas as pd

# Define the expected 4-year interval years
expected_years = set(range(2000, 2021, 4))  # {2000, 2004, ..., 2020}

# Filter just the columns we need
subset = unemployment[['State', 'County', 'Year', 'Unemployment rate']]

# Remove rows where 'Unemployment Rate' is missing
subset = subset.dropna(subset=['Unemployment rate'])

# Group by county and collect available years
years_present = subset.groupby(['State', 'County'])['Year'].apply(set).reset_index()

# Find counties missing any expected year
years_present['Missing Years'] = years_present['Year'].apply(lambda y: expected_years - y)
counties_with_missing_years = years_present[years_present['Missing Years'].apply(len) > 0]

# Display the result
print("Counties missing data for at least one year:")
print(counties_with_missing_years[['State', 'County', 'Missing Years']])
```
