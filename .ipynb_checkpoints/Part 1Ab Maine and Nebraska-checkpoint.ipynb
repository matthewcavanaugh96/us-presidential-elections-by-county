# Maine and Nebraska

The vast majority of states allocate all of their electoral votes to the statewide winner.

However, Maine and Nebraska use a different system, giving two statewide and one for each of its Congressional districts. It is therefore possible for each state to split its electoral votes. This happened to Maine in 2016 and 2020, and Nebraska in 2008 and 2020.

Since the Congressional districts do not precisely correspond with county lines, we will need to use some other method (likely Geospatial software) to reconcile the differing boundaries.

Our standard county dataset will be left intact for analysis and visualizations, but in simulating electoral vote totals by state, the data for ME and NE will have to be supplemented with the  Congressional district estimates, and a special allowance for this written into our summing function.
