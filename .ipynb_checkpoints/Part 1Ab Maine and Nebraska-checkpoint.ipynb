# Maine and Nebraska

The vast majority of states allocate all of their electoral votes to the statewide winner.

However, Maine and Nebraska use a different system, giving two statewide and one for each of its Congressional districts. It is therefore possible for each state to split its electoral votes, as happened to both in 2020 and 2024. 

Since the Congressional districts do not precisely correspond with county lines, we will need to use some other method (likely Geospatial software) to reconcile the differing boundaries.

Our standard county dataset will be left intact for analysis and visualizations, but in simulating electoral vote totals by state, the data for ME and NE will have to be dropped and replaced with our recalculated ones by Congressional district.
