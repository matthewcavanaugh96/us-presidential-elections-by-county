# Alaska

Alaska does not have counties in the traditional sense, but is instead broken into "organized" and "unorganized" boroughs. This would be fine except some data sources do not use these, but instead use "districts" with entirely different boundaries and names. This means that, while I could use past election data, it is not currently possible to merge with some other datasets. 

When or if I find a method to reconcile these differing statistical areas, I will outline my process here. Until then, I will have to exclude Alaska from the dataset.
