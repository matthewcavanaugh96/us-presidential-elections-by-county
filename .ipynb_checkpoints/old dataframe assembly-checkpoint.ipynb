# OLD 3P

```python
# Compile a DataFrame of third-party performance by county
# Group by State and County
elections_grouped = elections.groupby(['State', 'County'])

# Find the average, max and min values for each county
third_party = elections_grouped.agg(
    Average_Third_Party_Percentage=('Third Party Percentage', 'mean'),
    Largest_Third_Party_Percentage=('Third Party Percentage', 'max'),
    Smallest_Third_Party_Percentage=('Third Party Percentage', 'min')
).reset_index()

# Compute average excluding the single largest outlier
def average_excluding_max_outlier(group):
    values = group['Third Party Percentage'].values
    mean = values.mean()
    max_outlier_index = np.argmax(np.abs(values - mean))
    reduced_values = np.delete(values, max_outlier_index)
    return reduced_values.mean() if len(reduced_values) > 0 else np.nan
avg_no_max_outlier = elections_grouped.apply(average_excluding_max_outlier).reset_index(name='Average 3P % Excl. Largest Outlier')

# Compute average excluding the single smallest outlier
def average_excluding_min_outlier(group):
    values = group['Third Party Percentage'].values
    mean = values.mean()
    min_outlier_index = np.argmin(np.abs(values - mean))
    reduced_values = np.delete(values, min_outlier_index)
    return reduced_values.mean() if len(reduced_values) > 0 else np.nan
avg_no_min_outlier = elections_grouped.apply(average_excluding_min_outlier).reset_index(name='Average 3P % Excl. Smallest Outlier')

# Find which year had the maximum value
idx_max = elections_grouped['Third Party Percentage'].idxmax()
year_of_max = elections.loc[idx_max, ['State', 'County', 'Year']].rename(columns={'Year': 'Year of Largest 3P %'})

# Find which year had the minimum value
idx_min = elections_grouped['Third Party Percentage'].idxmin()
year_of_min = elections.loc[idx_min, ['State', 'County', 'Year']].rename(columns={'Year': 'Year of Smallest 3P %'})

# Merge
third_party = third_party.merge(year_of_max, on=['State', 'County'])
third_party = third_party.merge(year_of_min, on=['State', 'County'])
third_party = third_party.merge(avg_no_max_outlier, on=['State', 'County'])
third_party = third_party.merge(avg_no_min_outlier, on=['State', 'County'])

# Rename columns for cleanness
third_party.columns = [col.replace('_', ' ').replace('Percentage', '%') for col in third_party.columns]

third_party = third_party.rename(columns={
    'Average Third Party %': 'Average 3P %',
    'Largest Third Party %': 'Largest 3P %',
    'Smallest Third Party %': 'Smallest 3P %'
})

# Compute the effects of outliers on the average
third_party['3P% Absolute Effect of Largest Outlier'] = ((third_party['Average 3P %'] - third_party['Average 3P % Excl. Largest Outlier'])).abs()
third_party['3P% Absolute Effect of Smallest Outlier'] = ((third_party['Average 3P %'] - third_party['Average 3P % Excl. Smallest Outlier'])).abs()
third_party['3P% Total Effect of Extremes'] = (third_party['3P% Absolute Effect of Largest Outlier'] + third_party['3P% Absolute Effect of Smallest Outlier'])
#third_party['3P% Squared Total Effect of Outliers'] = third_party['3P% Total Effect of Outliers'] ** 2

# Compare the largest and smallest value to the county's own average
third_party['Largest 3P % ± County Average'] = third_party['Largest 3P %'] - third_party['Average 3P %']
third_party['Smallest 3P % ± County Average'] = third_party['Smallest 3P %'] - third_party['Average 3P %']

# Calculate the national average 3P
national_average_mean_3p = third_party['Average 3P %'].mean()
# Calculate each county's average 3P as compared to the national average
third_party['Average 3P % (± national avg)'] = third_party['Average 3P %'] - national_average_mean_3p

# Calculate the national average largest 3P
national_average_max_3p = third_party['Largest 3P %'].mean()
# Calculate each county's largest 3P as compared to the national average
third_party['Largest 3P % (± national avg)'] = third_party['Largest 3P %'] - national_average_max_3p

# Calculate the national average smallest 3P
national_average_min_3p = third_party['Smallest 3P %'].mean()
# Calculate each county's smallest 3P as compared to the national average
third_party['Smallest 3P % (± national avg)'] = third_party['Smallest 3P %'] - national_average_min_3p


third_party.head(5)
```

# old Swings and Volatility

```python
swings_and_volatility = pd.merge(county_swings, third_party, on=['State', 'County'])

# Calculate an adjusted volatility score
# Prevent infinite values
swings_and_volatility['Volatility div. by absolute change'].replace([np.inf, -np.inf], 0, inplace=True)
```

```python
swings_and_volatility['Average D-R Absolute Swing'] = ((
    swings_and_volatility['2000-2004 absolute swing'] +
    swings_and_volatility['2004-2008 absolute swing'] +
    swings_and_volatility['2008-2012 absolute swing'] +
    swings_and_volatility['2012-2016 absolute swing'] +
    swings_and_volatility['2016-2020 absolute swing']) / 5)


swings_and_volatility['Largest D-R Absolute Swing'] = swings_and_volatility[['2000-2004 absolute swing', '2004-2008 absolute swing', '2008-2012 absolute swing', '2012-2016 absolute swing', '2016-2020 absolute swing']].max(axis=1)
swings_and_volatility['Largest D-R Absolute Swing ± County Average'] = swings_and_volatility['Largest D-R Absolute Swing'] - swings_and_volatility['Largest D-R Absolute Swing'].mean()

swings_and_volatility['Smallest D-R Absolute Swing'] = swings_and_volatility[['2000-2004 absolute swing', '2004-2008 absolute swing', '2008-2012 absolute swing', '2012-2016 absolute swing', '2016-2020 absolute swing']].min(axis=1)
swings_and_volatility['Smallest D-R Absolute Swing ± County Average'] = swings_and_volatility['Smallest D-R Absolute Swing'] - swings_and_volatility['Smallest D-R Absolute Swing'].mean()

swings_and_volatility['Largest Minus Smallest D-R Swing'] = swings_and_volatility['Largest D-R Absolute Swing'] - swings_and_volatility['Smallest D-R Absolute Swing']
swings_and_volatility['Largest D-R Swing ± Natl. Average'] = swings_and_volatility['Largest D-R Absolute Swing'] - swings_and_volatility['Average D-R Absolute Swing'].mean()
swings_and_volatility['Smallest D-R Swing ± Natl. Average'] = swings_and_volatility['Smallest D-R Absolute Swing'] - swings_and_volatility['Average D-R Absolute Swing'].mean()
```

```python
# Compile two-party swing statistics by county

# Compute average excluding the single largest outlier
def average_swing_excluding_max_outlier(group):
    values = group['Third Party Percentage'].values
    mean = values.mean()
    max_outlier_index = np.argmax(np.abs(values - mean))
    reduced_values = np.delete(values, max_outlier_index)
    return reduced_values.mean() if len(reduced_values) > 0 else np.nan
avg_swing_no_max_outlier = elections_grouped.apply(average_swing_excluding_max_outlier).reset_index(name='Average D-R Excl. Largest Outlier')

# Compute average excluding the single smallest outlier
def average_swing_excluding_min_outlier(group):
    values = group['Third Party Percentage'].values
    mean = values.mean()
    min_outlier_index = np.argmin(np.abs(values - mean))
    reduced_values = np.delete(values, min_outlier_index)
    return reduced_values.mean() if len(reduced_values) > 0 else np.nan
avg_swing_no_min_outlier = elections_grouped.apply(average_swing_excluding_min_outlier).reset_index(name='Average D-R Excl. Smallest Outlier')


# Find which year had the maximum value
idx_max_dr = elections_grouped['Third Party Percentage'].idxmax()
year_of_max_dr = elections.loc[idx_max_dr, ['State', 'County', 'Year']].rename(columns={'Year': 'Year of Largest D-R Swing'})

# Find which year had the minimum value
idx_min_dr = elections_grouped['Third Party Percentage'].idxmin()
year_of_min_dr = elections.loc[idx_min_dr, ['State', 'County', 'Year']].rename(columns={'Year': 'Year of Smallest D-R Swing'})



swings_and_volatility = swings_and_volatility.merge(avg_swing_no_max_outlier, on=['State', 'County'])
swings_and_volatility = swings_and_volatility.merge(avg_swing_no_min_outlier, on=['State', 'County'])
swings_and_volatility = swings_and_volatility.merge(year_of_max_dr, on=['State', 'County'])
swings_and_volatility = swings_and_volatility.merge(year_of_min_dr, on=['State', 'County'])


swings_and_volatility['D-R Absolute Effect of Largest Outlier'] = ((swings_and_volatility['Average D-R Absolute Swing'] - swings_and_volatility['Average D-R Excl. Largest Outlier'])).abs()
swings_and_volatility['D-R Absolute Effect of Smallest Outlier'] = ((swings_and_volatility['Average D-R Absolute Swing'] - swings_and_volatility['Average D-R Excl. Smallest Outlier'])).abs()

swings_and_volatility['D-R Total Effect of Extremes'] = swings_and_volatility['D-R Absolute Effect of Largest Outlier'] + swings_and_volatility['D-R Absolute Effect of Smallest Outlier']
```
