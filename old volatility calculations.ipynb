# V3A

```python
# Penalize counties that swung hard one way, then reversed
swings_and_volatility['Reversion Penalty'] = (
    -np.minimum(
        abs(swings_and_volatility['2012-2016 swing']),
        abs(swings_and_volatility['2016-2020 swing'])
    ) * np.sign(
        swings_and_volatility['2012-2016 swing'] * swings_and_volatility['2016-2020 swing']
    )  # Only penalize if signs are opposite
)


swings_and_volatility['Adjusted volatility score v3a'] = (
    swings_and_volatility['Volatility'] + 
    swings_and_volatility['2016-2020 absolute swing'] + 
    swings_and_volatility['Volatility minus absolute change'] +
    (swings_and_volatility['Volatility div absolute change'] / 100) +
    swings_and_volatility['Average 3P % (± national avg)'] + 
    swings_and_volatility['Max 3P % (± national avg)'] + 
    swings_and_volatility['Min 3P % (± national avg)'] +
    swings_and_volatility['3P% Net Effect of Outliers'] +
    swings_and_volatility['2P Net Effect of Outliers'] -
    swings_and_volatility['Largest minus smallest'] +
    swings_and_volatility['Reversion Penalty']  # newly added mitigation
)
```

# V3B

```python
swings_and_volatility['Reversion Penalty'] = (
    -2 * np.minimum(
        abs(swings_and_volatility['2012-2016 swing']),
        abs(swings_and_volatility['2016-2020 swing'])
    ) * (swings_and_volatility['2012-2016 swing'] * swings_and_volatility['2016-2020 swing'] < 0)
)

swings_and_volatility['Recency-Weighted Swing'] = (
    swings_and_volatility['2000-2004 absolute swing'] * 0.5 +
    swings_and_volatility['2004-2008 absolute swing'] * 0.75 +
    swings_and_volatility['2008-2012 absolute swing'] * 1.0 +
    swings_and_volatility['2012-2016 absolute swing'] * 1.25 +
    swings_and_volatility['2016-2020 absolute swing'] * 1.5
)

swings_and_volatility['Adjusted volatility score v3B'] = (
    swings_and_volatility['Volatility'] + 
    swings_and_volatility['Recency-Weighted Swing'] + 
    swings_and_volatility['Volatility minus absolute change'] +
    (swings_and_volatility['Volatility div absolute change'] / 100) +
    swings_and_volatility['Average 3P % (± national avg)'] + 
    swings_and_volatility['Max 3P % (± national avg)'] + 
    swings_and_volatility['Min 3P % (± national avg)'] +
    swings_and_volatility['3P% Net Effect of Outliers'] +
    swings_and_volatility['2P Net Effect of Outliers'] -
    swings_and_volatility['Largest minus smallest'] +
    swings_and_volatility['Reversion Penalty']
)
```

```python
swings_and_volatility.nlargest(10, 'Adjusted volatility score v3B')
```

```python
# Calculate if swing and reversal occurred
swing_1 = swings_and_volatility['2012-2016 swing']
swing_2 = swings_and_volatility['2016-2020 swing']

# Check for reversal
opposite = (swing_1 * swing_2 < 0)

# How well they cancel each other
magnitude_similarity = 1 - abs(abs(swing_1) - abs(swing_2)) / (abs(swing_1) + abs(swing_2) + 1e-5)

# Apply a sharp penalty for high-cancellation reversals
swings_and_volatility['Isolated Spike Penalty'] = (
    -4 * np.minimum(abs(swing_1), abs(swing_2)) * opposite * magnitude_similarity
)

swings_and_volatility['Adjusted volatility score v3c'] = (
    swings_and_volatility['Volatility'] + 
    swings_and_volatility['Recency-Weighted Swing'] + 
    swings_and_volatility['Volatility minus absolute change'] +
    (swings_and_volatility['Volatility div absolute change'] / 100) +
    swings_and_volatility['Average 3P % (± national avg)'] + 
    swings_and_volatility['Max 3P % (± national avg)'] + 
    swings_and_volatility['Min 3P % (± national avg)'] +
    swings_and_volatility['3P% Net Effect of Outliers'] +
    swings_and_volatility['2P Net Effect of Outliers'] -
    swings_and_volatility['Largest minus smallest'] +
    swings_and_volatility['Isolated Spike Penalty']
)
```

```python
swings_and_volatility.nlargest(10, 'Adjusted volatility score v3c')
```

Something like this feels right. Many of these Utah counties did not sustain their 2016 shift, but in 2020 still ended up noticeably more competitive than pre-2016.
BUT - The swing reversal metric should be more robust to any year. While still keeping the recency bonus.

# Old 2PA Absolute Error calculation
Logic extended to other categories

```python
# Step 1: Compute the national average of Absolute Error (2PA)
national_avg_error = rfr_2pas['Absolute Error (2PA)'].mean()

# Step 2: Define function to compute all stats
def summarize_with_offsets(group):
    errors = group['Absolute Error (2PA)'].values
    years = group['Year'].values
    
    largest_idx = np.argmax(errors)
    smallest_idx = np.argmin(errors)
    
    largest = errors[largest_idx]
    smallest = errors[smallest_idx]
    mean = errors.mean()
    
    # Exclude extremes
    errors_excl_largest = np.delete(errors, largest_idx)
    errors_excl_smallest = np.delete(errors, smallest_idx)
    mean_excl_largest = errors_excl_largest.mean() if len(errors_excl_largest) > 0 else np.nan
    mean_excl_smallest = errors_excl_smallest.mean() if len(errors_excl_smallest) > 0 else np.nan

    return pd.Series({
        'Absolute Error (2PA) County Avg': mean,
        'Absolute Error (2PA) Largest': errors[largest_idx],
        'Absolute Error (2PA) Largest ± County Avg': largest - mean,
        'Absolute Error (2PA) Year of Largest': years[largest_idx],
        'Absolute Error (2PA) Smallest': errors[smallest_idx],
        'Absolute Error (2PA) Smallest ± County Avg': smallest - mean,
        'Absolute Error (2PA) Year of Smallest': years[smallest_idx],
        'Absolute Error (2PA) Average ± Natl. Avg': mean - national_avg_error,
        'Absolute Error (2PA) Largest ± Natl. Avg': largest - national_avg_error,
        'Absolute Error (2PA) Smallest ± Natl. Avg': smallest - national_avg_error,
        'Absolute Error (2PA) Avg. Excl. Largest Outlier': mean_excl_largest,
        'Absolute Error (2PA) Avg. Excl. Smallest Outlier': mean_excl_smallest,
        'Absolute Error (2PA) Effect of Largest Error': abs(mean - mean_excl_largest),
        'Absolute Error (2PA) Effect of Smallest Error': abs(mean - mean_excl_smallest),
        'Absolute Error (2PA) Total Effect of Extremes': abs(mean - mean_excl_largest) + abs(mean - mean_excl_smallest)
    })


# Apply function to grouped DataFrame
summary_df = rfr_2pas.groupby(['State', 'County']).apply(summarize_with_offsets).reset_index()

# Set columns to integer
summary_df['Absolute Error (2PA) Year of Largest'] = summary_df['Absolute Error (2PA) Year of Largest'].astype(int)
summary_df['Absolute Error (2PA) Year of Smallest'] = summary_df['Absolute Error (2PA) Year of Smallest'].astype(int)
```
